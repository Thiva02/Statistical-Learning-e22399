**1. Proof of Maximum Likelihood Estimator Bias and Bessel's Correction**

To demonstrate that the raw MLE covariance matrix $\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}$ is biased, we expand its expectation. The MLE estimator is defined as:
$$\widehat{\boldsymbol{\Sigma}}_{\text{MLE}} = \frac{1}{n} \sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T$$

By adding and subtracting the true population mean $\boldsymbol{\mu}$, we can decompose the inner term:
$$(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n) = (\mathbf{X}_i - \boldsymbol{\mu}) - (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})$$

Expanding the quadratic form and summing over all $n$ observations yields:
$$\sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T = \sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T - n(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T$$

Taking the expectation $\mathbb{E}[\cdot]$ of both sides:
* The expectation of the first term is $n\boldsymbol{\Sigma}$ (since each observation contributes $\boldsymbol{\Sigma}$).
* The expectation of the second term relates to the variance of the sample mean, which is $n(\frac{1}{n}\boldsymbol{\Sigma}) = \boldsymbol{\Sigma}$.

Subtracting these gives the expected sum of squared deviations:
$$\mathbb{E} \left[ \sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T \right] = n\boldsymbol{\Sigma} - \boldsymbol{\Sigma} = (n-1)\boldsymbol{\Sigma}$$

Multiplying by the MLE prefix $\frac{1}{n}$, we get:
$$\mathbb{E}[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}] = \frac{n-1}{n}\boldsymbol{\Sigma}$$
Because $\frac{n-1}{n}\boldsymbol{\Sigma} \neq \boldsymbol{\Sigma}$, the MLE estimator fundamentally underestimates the true population covariance and is **biased**.

**Bessel's Correction:**
To correct this underestimation and yield an unbiased estimator, we apply Bessel's correction by multiplying $\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}$ by $\frac{n}{n-1}$:
$$\mathbf{S} = \frac{n}{n-1} \widehat{\boldsymbol{\Sigma}}_{\text{MLE}} = \frac{1}{n-1} \sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T$$
Taking the expectation proves it is unbiased:
$$\mathbb{E}[\mathbf{S}] = \frac{1}{n-1} \mathbb{E} \left[ \dots \right] = \frac{1}{n-1} (n-1)\boldsymbol{\Sigma} = \boldsymbol{\Sigma}$$

**2. Type I and Type II Errors in Structural Health Monitoring**

* **Type I Error ($\alpha$):** A False Alarm. This occurs when the diagnostic layer incorrectly rejects the null hypothesis, triggering an alert that structural damage exists when the asset is actually perfectly healthy.
* **Type II Error ($\beta$):** A Missed Detection. This occurs when the diagnostic layer fails to reject the null hypothesis, reporting that the structure is healthy when a physical fault or internal fracture is actively occurring.

**Mathematical Consequence of Ultra-Conservative $\alpha = 0.0001$:**
Setting an ultra-low threshold for false alarms forces the statistical decision boundary (the critical threshold value) to be pushed drastically outward into the extreme tails of the distribution.
* **Statistical Power Loss:** Because Power is defined as $1 - \beta$, decreasing $\alpha$ inevitably increases $\beta$. The diagnostic layer loses statistical power and becomes highly insensitive to actual physical anomalies.
* **Geometric Consequence:** The geometric volume of the "healthy operation" confidence ellipsoid expands enormously. It becomes so large that it swallows the distributions of actual damaged states, allowing active structural drift to comfortably hide inside the "healthy" boundaries.


**1. Slutsky's Theorem (Rigorous Statement)**
Let $\mathbf{X}_n$ and $\mathbf{Y}_n$ be sequences of random vectors. If $\mathbf{X}_n$ converges in distribution to a random vector $\mathbf{X}$ ($\mathbf{X}_n \xrightarrow{d} \mathbf{X}$) and $\mathbf{Y}_n$ converges in probability to a constant matrix $\mathbf{c}$ ($\mathbf{Y}_n \xrightarrow{p} \mathbf{c}$), then:
1. $\mathbf{X}_n + \mathbf{Y}_n \xrightarrow{d} \mathbf{X} + \mathbf{c}$
2. $\mathbf{Y}_n \mathbf{X}_n \xrightarrow{d} \mathbf{c}\mathbf{X}$
3. $\mathbf{Y}_n^{-1} \mathbf{X}_n \xrightarrow{d} \mathbf{c}^{-1}\mathbf{X}$ (provided $\mathbf{c}$ is invertible).

**2. Operationalizing the Parametric Model**
The Lindeberg-Lévy CLT establishes that:
$$\sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \xrightarrow{d} \mathscr{N}(\mathbf{0}, \boldsymbol{\Sigma})$$

However, $\boldsymbol{\Sigma}$ is unobservable. By the Weak Law of Large Numbers, the empirical covariance $\mathbf{S}$ converges in probability to the true covariance $\boldsymbol{\Sigma}$ ($\mathbf{S} \xrightarrow{p} \boldsymbol{\Sigma}$). Therefore, any continuous mapping also converges, meaning $\mathbf{S}^{-1/2} \xrightarrow{p} \boldsymbol{\Sigma}^{-1/2}$.

We wish to evaluate the standardized distribution using our empirical estimator:
$$\sqrt{n} \mathbf{S}^{-1/2} (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})$$

By applying Slutsky's Theorem (property 2), we multiply the sequence converging in probability ($\mathbf{S}^{-1/2}$) with the sequence converging in distribution:
$$\mathbf{S}^{-1/2} \left[ \sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \right] \xrightarrow{d} \boldsymbol{\Sigma}^{-1/2} \mathscr{N}(\mathbf{0}, \boldsymbol{\Sigma}) = \mathscr{N}(\mathbf{0}, \mathbf{I})$$

Because the asymptotic limit resolves to the exact same standard normal distribution regardless of whether we use $\mathbf{S}$ or $\boldsymbol{\Sigma}$, Slutsky's Theorem provides the rigorous justification to substitute $\mathbf{S}$ into the covariance parameter. Thus, for sufficiently large finite sample sizes, we can operationalize the model as:
$$\widehat{\boldsymbol{\mu}}_n \sim \mathscr{N}\left({\boldsymbol{\mu}}, \frac{1}{n}\mathbf{S}\right)$$


```python
import numpy as np
import pandas as pd
import scipy.stats as stats

# --- DO NOT MODIFY: Synthetic Strain Sensor Data Generator ---
np.random.seed(42)
n_samples = 5000
n_features = 3

# Generating base stationary multivariate normal noise
base_data = np.random.multivariate_normal(
    mean=[0.5, -0.2, 1.1],
    cov=[[0.09, 0.02, 0.01],
         [0.02, 0.06, 0.03],
         [0.01, 0.03, 0.05]],
    size=n_samples
)

# Injecting a subtle, hidden structural drift in the final 1,000 snapshots
base_data[4000:, 0] += 0.015  # Sensor 1 drift
base_data[4000:, 2] -= 0.010  # Sensor 3 drift

df_strain = pd.DataFrame(base_data, columns=['Strain_Ch1', 'Strain_Ch2', 'Strain_Ch3'])
# ----------------------------------------------------------------

def verify_first_moment_homogeneity(df: pd.DataFrame, g_chunks: int = 5) -> dict:
    """
    Partitions the dataset into g chunks and evaluates first-moment homogeneity
    via Wilks' Lambda and Bartlett's Chi-Square asymptotic transformation.

    Author: Thivarshikan Yogaretnam (E/22/384)
    """
    n = len(df)
    m = df.shape[1]
    n_j = n // g_chunks

    data_array = df.values
    global_mean = np.mean(data_array, axis=0)

    # Initialize Within-Chunk (W) and Between-Chunk (B) covariance matrices
    W = np.zeros((m, m))
    B = np.zeros((m, m))

    # 1 & 2. Split into blocks and compute matrices
    for i in range(g_chunks):
        chunk = data_array[i * n_j : (i + 1) * n_j]
        chunk_mean = np.mean(chunk, axis=0)

        # Calculate W (sum of squares within the chunk)
        centered_chunk = chunk - chunk_mean
        W += centered_chunk.T @ centered_chunk

        # Calculate B (variance between chunk mean and global mean)
        mean_diff = (chunk_mean - global_mean).reshape(m, 1)
        B += n_j * (mean_diff @ mean_diff.T)

    # 3. Compute Wilks' Lambda
    det_W = np.linalg.det(W)
    det_T = np.linalg.det(W + B)
    wilks_lambda = det_W / det_T

    # 4. Apply Bartlett's Chi-Square approximation
    chi2_calc = -(n - 1 - (m + g_chunks) / 2) * np.log(wilks_lambda)

    # Calculate Degrees of Freedom and p-value
    df_chi2 = m * (g_chunks - 1)
    p_value = stats.chi2.sf(chi2_calc, df_chi2)

    # Diagnostic Conclusion based on alpha = 0.05
    if p_value < 0.05:
        conclusion = "Drift Detected (Reject H0) - First moment is NOT homogeneous."
    else:
        conclusion = "Stationary (Fail to Reject H0) - First moment remains homogeneous."

    return {
        "Wilks_Lambda": round(wilks_lambda, 6),
        "Bartlett_Chi2": round(chi2_calc, 2),
        "Degrees_of_Freedom": df_chi2,
        "p_value": p_value,
        "Conclusion": conclusion
    }

# Execute your completed function below:
results = verify_first_moment_homogeneity(df_strain, g_chunks=5)

print("--- DIAGNOSTIC RESULTS ---")
for key, value in results.items():
    print(f"{key}: {value}")
```
